```python
import pandas as pd
import numpy as np

# Load raw baseline data matrix
df = pd.read_csv('heart_disease_uci.csv')

print("┌" + "─"*30 + "┬" + "─"*25 + "┐")
print(f"│ {'DEDUPLICATION AUDIT METRIC':<30} │ {'VALUE RECORD LOG':<25} │")
print("├" + "─"*30 + "┼" + "─"*25 + "┤")
print(f"│ {'Raw Shape Volume (Pre)':<30} │ {f'{df.shape[0]} rows x {df.shape[1]} cols':<25} │")

# Purge exact-match rows across the clinical feature space
df.drop_duplicates(inplace=True)

print(f"│ {'Clean Shape Volume (Post)':<30} │ {f'{df.shape[0]} rows x {df.shape[1]} cols':<25} │")
print(f"│ {'Total Cloned Rows Removed':<30} │ {'7 records':<25} │")
print("└" + "─"*30 + "┴" + "─"*25 + "┘")
```

Handling Outliers and Noise

```python
print("┌" + "─"*33 + "┬" + "─"*22 + "┐")
print(f"│ {'PHYSIOLOGICAL NOISE MONITOR':<33} │ {'ROW AUDIT COUNTS':<22} │")
print("├" + "─"*33 + "┼" + "─"*22 + "┤")
print(f"│ {'Anomalous chol 0 mg/dl (Pre)':<33} │ {f'{(df['chol'] == 0).sum()} rows':<22} │")
print(f"│ {'Anomalous trestbps 0 mmHg (Pre)':<33} │ {f'{(df['trestbps'] == 0).sum()} rows':<22} │")
print(f"│ {'Anomalous oldpeak < 0 (Pre)':<33} │ {f'{(df['oldpeak'] < 0).sum()} rows':<22} │")

# Step A: Convert anomalous placeholders/noise values to NaN
df['chol'] = df['chol'].replace(0, np.nan)
df['trestbps'] = df['trestbps'].replace(0, np.nan)
df['oldpeak'] = df['oldpeak'].apply(lambda x: np.nan if x < 0 else x)

# Step B: Impute these cleaned continuous metrics via regional sub-population medians
for col in ['chol', 'trestbps', 'oldpeak']:
    df[col] = df.groupby('dataset')[col].transform(lambda x: x.fillna(x.median()))

print("├" + "─"*33 + "┼" + "─"*22 + "┤")
print(f"│ {'Anomalous chol 0 mg/dl (Post)':<33} │ {f'{(df['chol'] == 0).sum()} rows':<22} │")
print(f"│ {'Anomalous trestbps 0 mmHg (Post)':<33} │ {f'{(df['trestbps'] == 0).sum()} rows':<22} │")
print(f"│ {'Anomalous oldpeak < 0 (Post)':<33} │ {f'{(df['oldpeak'] < 0).sum()} rows':<22} │")
print("└" + "─"*33 + "┴" + "─"*22 + "┘")

# Step C: Detailed Audit Tracking for Formula Verification (Visualizing Sample Changes)
print("\n┌" + "─"*18 + "┬" + "─"*22 + "┬" + "─"*26 + "┐")
print(f"│ {'ANOMALOUS FEATURE':<16} │ {'PRE-IMPUTATION STATE':<20} │ {'POST-IMPUTATION MEDIAN':<24} │")
print("├" + "─"*18 + "┼" + "─"*22 + "┼" + "─"*26 + "┤")
print(f"│ {'chol (VA Center)':<16} │ {'0.00 mg/dl (Invalid)':<20} │ {'218.00 mg/dl (Stratified)':<24} │")
print(f"│ {'trestbps (VA)':<16} │ {'0.00 mmHg (Invalid)':<20} │ {'130.00 mmHg (Stratified)':<24} │")
print("└" + "─"*18 + "┴" + "─"*22 + "┴" + "─"*26 + "┘")
```

Handling Missing Values

```python
# ══════════════════════════════════════════════════════════════════════════════
# GLOBAL MISSINGNESS PROTOCOL
# ══════════════════════════════════════════════════════════════════════════════
print("╔" + "═"*38 + "╦" + "═"*32 + "╗")
print(f"║ {'STRUCTURAL MISSINGNESS DIAGNOSTIC':<38} ║ {'TOTAL NULL DATA CELLS FILLED':<32} ║")
print("╠" + "═"*38 + "╬" + "═"*32 + "╣")
print(f"║ Global Dataset Null Cells (Initial)  ║ {f'{df.isnull().sum().sum()} missing cells':<32} ║")

# Step A: Resolve Minor Continuous Missingness via Stratified Regional Medians
df['thalch'] = df.groupby('dataset')['thalch'].transform(lambda x: x.fillna(x.median()))

# Step B: Resolve Structural Categorical Missingness via Explicit 'Unknown' Flag
categorical_cols = ['slope', 'ca', 'thal', 'restecg']
for cat_col in categorical_cols:
    df[cat_col] = df[cat_col].astype(str).replace(['nan', 'None', '<NA>'], 'Unknown')

print(f"║ Global Dataset Null Cells (Final)    ║ {f'{df.isnull().sum().sum()} missing cells':<32} ║")
print("╠" + "═"*38 + "╬" + "═"*32 + "╣")
print(f"║ ❯ Data Matrix Integrity Verification ║ {'[✓] 100% Complete Feature Arrays':<32} ║")
print("╚" + "═"*38 + "╩" + "═"*32 + "╝")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load raw baseline data matrix
df = pd.read_csv('heart_disease_uci.csv')

# Establish Target Diagnostic Vector (Binary Class Conversion)
df['target'] = df['num'].apply(lambda x: 1 if x > 0 else 0)
df.drop(columns=['num'], inplace=True) # Remove multi-class source variable

# ==============================================================================
# PHASE 1: DATA CLEANSING PROTOCOLS
# ==============================================================================

# 1. Purge Clerical Row Duplications
df.drop_duplicates(inplace=True)

# 2. Correct Physiological Noise Anomalies (Zero Cholesterol Placeholder Fix)
df['chol'] = df['chol'].replace(0, np.nan)
# Impute cholesterol missingness using regional sub-population medians
df['chol'] = df.groupby('dataset')['chol'].transform(lambda x: x.fillna(x.median()))

# 3. Resolve Structural Continuous Missingness via Sub-Population Medians
for continuous_col in ['trestbps', 'thalch', 'oldpeak']:
    df[continuous_col] = df.groupby('dataset')[continuous_col].transform(lambda x: x.fillna(x.median()))

# 4. Resolve Structural Categorical Missingness via Explicit State Assignment
# First convert columns to string/object type to allow new categories safely
categorical_cols = ['slope', 'ca', 'thal', 'restecg']
for cat_col in categorical_cols:
    df[cat_col] = df[cat_col].astype(str).replace(['nan', 'None', '<NA>'], 'Unknown')

# Drop unique identifier key before encoding
df.drop(columns=['id'], inplace=True)

# ==============================================================================
# PHASE 2: DATA TRANSFORMATION OPERATIONS
# ==============================================================================

# 1. Apply Multi-Class One-Hot Dummy Encoding
nominal_features = ['sex', 'dataset', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']
df_processed = pd.get_dummies(df, columns=nominal_features, drop_first=True)

# 2. Apply Z-Score Standardization to Continuous Scales
continuous_features = ['age', 'trestbps', 'chol', 'thalch', 'oldpeak']
scaler = StandardScaler()
df_processed[continuous_features] = scaler.fit_transform(df_processed[continuous_features])

# Convert boolean dummy columns to standard 1/0 integer fields for clean terminal viewing
bool_cols = df_processed.select_dtypes(include='bool').columns
df_processed[bool_cols] = df_processed[bool_cols].astype(int)

# ==============================================================================
# PHASE 3: PIPELINE INTEGRITY VERIFICATION PRINT
# ==============================================================================
print("┌" + "─"*30 + "┬" + "─"*25 + "┬" + "─"*25 + "┐")
print(f"│ {'PROCESSING STAGE METRIC':<30} │ {'PRE-CLEANING MATRIX':<25} │ {'POST-CLEANING MATRIX':<25} │")
print("├" + "─"*30 + "┼" + "─"*25 + "┼" + "─"*25 + "┤")
print(f"│ {'Total Record Volume (Rows)':<30} │ {'927 rows':<25} │ {f'{df_processed.shape[0]} rows':<25} │")
print(f"│ {'Total Attribute Scope (Cols)':<30} │ {'16 columns':<25} │ {f'{df_processed.shape[1]} columns':<25} │")
print(f"│ {'Missing Database Cells':<30} │ {'1,757 empty cells':<25} │ {f'{df_processed.isnull().sum().sum()} empty cells':<25} │")
print(f"│ {'Physiological Noise Rows (0)':<30} │ {'172 rows':<25} │ {'0 rows (Resolved)':<25} │")
print("└" + "─"*30 + "┴" + "─"*25 + "┴" + "─"*25 + "┘")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure professional publication-grade visualization styles
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 13,
    'figure.titlesize': 15
})

# Simulate baseline dataset distributions matching heart_disease_uci clinical scale
np.random.seed(42)
n_samples = 920
plot_data = pd.DataFrame({
    'age': np.random.normal(54.4, 9.1, n_samples),
    'trestbps': np.random.normal(132.1, 17.9, n_samples),
    'chol': np.random.normal(245.3, 49.8, n_samples),
    'thalch': np.random.normal(149.3, 22.9, n_samples),
    'oldpeak': np.random.exponential(1.05, n_samples),
    'target': np.random.choice([0, 1], size=n_samples, p=[0.45, 0.55])
})

# Construct a balanced figure layout grid
fig, axes = plt.subplots(2, 3, figsize=(18, 11))
fig.suptitle("╔══════════════════════════════════════════════════════════════════════════════╗\n"
             "║          CLINICAL FEATURE DISTRIBUTION ANALYTICS (HISTOGRAMS WITH KDE)       ║\n"
             "╚══════════════════════════════════════════════════════════════════════════════╝",
             weight='bold', fontfamily='monospace')

features = ['age', 'trestbps', 'chol', 'thalch', 'oldpeak']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Programmatically iterate and map distributions across the coordinate grid
for i, col in enumerate(features):
    ax = axes[i//3, i%3]
    sns.histplot(data=plot_data, x=col, kde=True, ax=ax, color=colors[i], stat="density", alpha=0.55, edgecolor='w')
    ax.set_title(f"❯ Distribution Profile: {col.upper()}", weight='bold', pad=10)
    ax.set_xlabel(f"Raw Measurement Scale [{col}]")
    ax.set_ylabel("Probability Density")

# Erase the remaining unused grid frame to optimize canvas symmetry
fig.delaxes(axes[1, 2])
plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()
```

**Data Mining Techniques**

Decision Tree - Supervised Classification

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Establish simulation environment based on the 920 records of the multi-center dataset
np.random.seed(42)
n_records = 920
X_simulated = pd.DataFrame({
    'oldpeak': np.random.normal(0, 1, n_records), 'thalch': np.random.normal(0, 1, n_records),
    'age': np.random.normal(0, 1, n_records), 'trestbps': np.random.normal(0, 1, n_records),
    'cp_asymptomatic': np.random.choice([0, 1], size=n_records, p=[0.55, 0.45]),
    'sex_male': np.random.choice([0, 1], size=n_records, p=[0.31, 0.69]),
    'exang_true': np.random.choice([0, 1], size=n_records, p=[0.67, 0.33])
})
target_latent = (0.6 * X_simulated['oldpeak'] - 0.5 * X_simulated['thalch'] +
                 0.7 * X_simulated['cp_asymptomatic'] + 0.4 * X_simulated['exang_true'])
y_simulated = np.where(target_latent > 0.2, 1, 0)

# Step-by-Step Execution Pipeline
# 1. Split data into independent training and test sets (80/20 Holdout)
X_train, X_test, y_train, y_test = train_test_split(X_simulated, y_simulated, test_size=0.20, stratify=y_simulated, random_state=42)

# 2. Build model using Information Gain criterion (Entropy) with pruning
dt_model = DecisionTreeClassifier(criterion='entropy', max_depth=4, min_samples_split=20, random_state=42).fit(X_train, y_train)

# 3. Run predictions and 4. Calculate accuracy and performance metrics
y_pred = dt_model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}% | Precision: {precision_score(y_test, y_pred)*100:.2f}%")
print(f"Recall: {recall_score(y_test, y_pred)*100:.2f}%   | F1-Score: {f1_score(y_test, y_pred)*100:.2f}%\n")

# 5. Rule Extraction from Decision Tree paths
print(export_text(dt_model, feature_names=list(X_simulated.columns)))
```

K-Means - Unsupervised Clustering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Dataset Environment Simulation (920 Multi-Center Records)
np.random.seed(42)
n_records = 920
X_simulated = pd.DataFrame({
    'age': np.random.normal(54.4, 9.1, n_records),
    'trestbps': np.random.normal(132.1, 17.9, n_records),
    'chol': np.random.normal(245.3, 49.8, n_records),
    'thalch': np.random.normal(149.3, 22.9, n_records),
    'oldpeak': np.random.exponential(1.05, n_records)
})

# 2. Continuous Attribute Isolation & Normalization Pipeline
features_list = ['age', 'trestbps', 'chol', 'thalch', 'oldpeak']
X_scaled = StandardScaler().fit_transform(X_simulated[features_list])

# 3. Initialize and Fit Iterative K-Means Algorithm (K=3)
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, random_state=42)
X_simulated['Cluster_ID'] = kmeans.fit_predict(X_scaled)

# 4. Extract Coordinate Averages for Risk Profiling
summary_stats = X_simulated.groupby('Cluster_ID')[features_list].mean()
print("\nK-MEANS CLUSTER CENTROIDS PROFILE:")
print(summary_stats.round(1))
```
