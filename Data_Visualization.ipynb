**3.0 DATASET CHARACTERISTICS**

Mean. Median. Mode, and Std. Deviation

```python
import pandas as pd

# Load your CSV
df = pd.read_csv("globalAirQuality.csv", low_memory=False)
df.columns = [c.strip() for c in df.columns]

# Select cities
selected_cities = ["Kuala Lumpur","Dubai","Cairo","Tehran","Toronto",
                   "Cape Town","Chennai","Tokyo","Beijing","Manila"]

df['city'] = df['city'].astype(str).str.strip()
df_sel = df[df['city'].isin(selected_cities)]

# Select NUMERICAL columns only
num_cols = [c for c in df_sel.columns if pd.api.types.is_numeric_dtype(df_sel[c])]

# Create a table for stats
stats_list = []

for col in num_cols:
    s = pd.to_numeric(df_sel[col], errors='coerce').dropna()
    if len(s) == 0:
        continue

    stats_list.append({
        "variable": col,
        "count": len(s),
        "mean": s.mean(),
        "median": s.median(),
        "mode": s.mode().iloc[0] if not s.mode().empty else None,
        "std": s.std()
    })

stats_df = pd.DataFrame(stats_list)

# DISPLAY RESULT
stats_df
```

**4.0 RELATIONSHP INVESTIGATION**
*   NO₂ vs PM10
*   O₃ vs Temperature
*   CO vs Humidity





```python
# Install seaborn if not already
!pip install seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Mount Google Drive if CSV is stored there (optional)
# from google.colab import drive
# drive.mount('/content/drive')

# ===============================
# Load the dataset
# Replace the path with your CSV location
df = pd.read_csv("/content/globalAirQuality.csv", low_memory=False)
df.columns = [c.strip() for c in df.columns]

# Select cities
selected_cities = ["Kuala Lumpur","Dubai","Cairo","Tehran","Toronto",
                   "Cape Town","Chennai","Tokyo","Beijing","Manila"]
df['city'] = df['city'].astype(str).str.strip()
df = df[df['city'].isin(selected_cities)].copy()

# Identify numeric columns
num_cols = [c for c in df.columns if df[c].dtype in ['float64','int64']]

# Optional: define columns explicitly if your dataset has slightly different names
pm10_col = [c for c in num_cols if 'pm10' in c.lower()][0]
no2_col = [c for c in num_cols if 'no2' in c.lower()][0]
o3_col = [c for c in num_cols if 'o3' in c.lower()][0]
co_col = [c for c in num_cols if 'co' in c.lower()][0]
temp_col = [c for c in num_cols if 'temp' in c.lower()][0]
humidity_col = [c for c in num_cols if 'humid' in c.lower()][0]

# ===============================
# Set seaborn style
sns.set(style="whitegrid")

# 1. NO2 vs PM10
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x=pm10_col, y=no2_col, hue='city', alpha=0.7)
sns.regplot(data=df, x=pm10_col, y=no2_col, scatter=False, color='black')
plt.title("NO2 vs PM10 (Selected Cities)")
plt.xlabel("PM10")
plt.ylabel("NO2")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

# 2. O3 vs Temperature
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x=temp_col, y=o3_col, hue='city', alpha=0.7)
sns.regplot(data=df, x=temp_col, y=o3_col, scatter=False, color='black')
plt.title("O3 vs Temperature (Selected Cities)")
plt.xlabel("Temperature")
plt.ylabel("O3")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

# 3. CO vs Humidity
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x=humidity_col, y=co_col, hue='city', alpha=0.7)
sns.regplot(data=df, x=humidity_col, y=co_col, scatter=False, color='black')
plt.title("CO vs Humidity (Selected Cities)")
plt.xlabel("Humidity")
plt.ylabel("CO")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
